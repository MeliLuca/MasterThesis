```python
# Generic config object
class Config():
    VOC_YEAR = '2007'
    VOC_TRAIN_PATH = '/VOC_train'
    VOC_TRAINVAL_PATH = '/VOC_trainval'
    VOC_TEST_PATH = '/VOC_test'
    DATASET_PATH = './dataset'
    SAVING_PATH = './models'

# Neural Network base model config object
class BaseModelConfig():
    patch_dim = 96
    gap = 48
    jitter = 7
    batch_size = 64
    num_epochs = 300
    lr = 0.0001
    
```

```python
# imports 
import os, time, math, random, json

from PIL import Image

import copy

# torch imports
import torch
from torchvision import transforms
import torch.optim as optim

# torchvision
import torchvision

# cv2
import cv2

# numpy
import numpy as np

# matplotlib
import matplotlib.pyplot as plt

from PIL import Image

# pandas
import pandas as pd

# glob (for path resolution with regexp)
from glob import glob

# sklearn
from sklearn.model_selection import StratifiedShuffleSplit, ShuffleSplit

# skimage
import skimage
import skimage.transform
from skimage import img_as_ubyte, img_as_float32

# tqdm (for outputting a progress bar)
from tqdm import tqdm

# init variables
np.random.seed(104)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
if not os.path.exists(Config.SAVING_PATH):
    os.makedirs(Config.SAVING_PATH)

if not os.path.exists(Config.DATASET_PATH):
    os.makedirs(Config.DATASET_PATH)   
if not os.path.exists(f'{Config.DATASET_PATH}/{Config.VOC_TRAIN_PATH}'):
    os.makedirs(f'{Config.DATASET_PATH}/{Config.VOC_TRAIN_PATH}')    
if not os.path.exists(f'{Config.DATASET_PATH}/{Config.VOC_TRAINVAL_PATH}'):
    os.makedirs(f'{Config.DATASET_PATH}/{Config.VOC_TRAINVAL_PATH}')  
if not os.path.exists(f'{Config.DATASET_PATH}/{Config.VOC_TEST_PATH}'):
    os.makedirs(f'{Config.DATASET_PATH}/{Config.VOC_TEST_PATH}')

train_data = torchvision.datasets.VOCDetection(f'{Config.DATASET_PATH}/{Config.VOC_TRAIN_PATH}', Config.VOC_YEAR, 'train', True)
print(len(train_data))

trainval_data = torchvision.datasets.VOCDetection(f'{Config.DATASET_PATH}/{Config.VOC_TRAINVAL_PATH}', Config.VOC_YEAR, 'trainval', True)
print(len(trainval_data))

test_data = torchvision.datasets.VOCDetection(f'{Config.DATASET_PATH}/{Config.VOC_TEST_PATH}', Config.VOC_YEAR, 'test', True)
print(len(test_data))
```

```python
def resize_image(image, pad = 0):
    im = np.array(image)
    im = im.copy()
    
    min_dim = (BaseModelConfig.patch_dim * 3) + (BaseModelConfig.gap * 2) + 2 + (BaseModelConfig.jitter * 2) + pad * 2
    
    if im.shape[0] < min_dim or im.shape[1] < min_dim:
        # resize image
        ratio1 = float(min_dim)/float(im.shape[0])
        ratio2 = float(min_dim)/float(im.shape[1])
        
        ratio = max(ratio1, ratio2)
        im = cv2.resize(im,(int(im.shape[0]*ratio), int(im.shape[1]*ratio)))
    
    immax=float(im.shape[0]*im.shape[1])
    
    if immax > 450000:
        return im
    
    maxdim=random.randint(immax,450000 + 1)
    ratio=float(maxdim)/float(immax);
    im = cv2.resize(im,(int(im.shape[0]*np.sqrt(ratio)), int(im.shape[1]*np.sqrt(ratio))))
    return im

def plot_image(img,text=None,should_save=False):
    plt.figure(figsize=(10, 10))
    npimg = img.numpy()
    plt.axis("off")
    if text:
        plt.text(75, 8, text, style='italic',fontweight='bold',
            bbox={'facecolor':'white', 'alpha':0.8, 'pad':10})
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    plt.show()  

def show_plot(iteration,loss,fname):
    plt.plot(iteration,loss)
    plt.savefig(fname)
    plt.show()
    
def display_canvas(patch_bucket, coordinates, title):
    x = np.linspace(8, 8, 100)
    y = np.cos(x)

    fig, ax = plt.subplots()

    idx = 0
    for i in reversed(range(3)):
        for j in range(0, 9, 3):
            ax.imshow(patch_bucket[idx], extent=[coordinates[i+j][0], 
                                          coordinates[i+j][0]+BaseModelConfig.patch_dim, 
                                          coordinates[i+j][1], 
                                          coordinates[i+j][1]+BaseModelConfig.patch_dim], aspect='auto')
            idx += 1
    ax.plot(x, y)
    plt.title(title)
    plt.show()

class UnNormalize(object):
    def __init__(self, mean, std):
        self.mean = mean
        self.std = std

    def __call__(self, tensor):
        for t, m, s in zip(tensor, self.mean, self.std):
            t.mul_(s).add_(m)
        return tensor
    
    
unorm = UnNormalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225))
```

```python
#############################
# Creating training dataset
#############################

training_set_list = []

for idx in range(0, len(train_data)):
    img, label = train_data[idx]
    img = resize_image(img)
    
    training_set_list.append([img])

print(f'Total training set elements: {len(training_set_list)}')

dataframe = pd.DataFrame(data=training_set_list, columns=['image'])

df_trn = dataframe
df_trn.head()
```

```python
#############################
# Creating validation dataset
#############################

validation_set_list = []

for idx in range(0, len(trainval_data)):
    img, label = trainval_data[idx]
    img = resize_image(img)
    
    validation_set_list.append([img])
    
print(f'Total validation set elements: {len(validation_set_list)}')
        
dataframe = pd.DataFrame(data=validation_set_list, columns=['image'])

df_val = dataframe
df_val.head()
```

```python
#########################################
# This class generates patches for training
#########################################

class OurDataset(torch.utils.data.Dataset):
    def __init__(self, patch_dim, gap, df, validate, transform=None, transform2 = None):
        self.patch_dim, self.gap = patch_dim, gap
        self.transform = transform
        self.transform2 = transform2

        if validate:
            self.train_data = df.values
        else:
            self.train_data = df.values
        
    def get_patch_from_grid(self, image, patch_dim, gap):        
        offset_y, offset_x = image.shape[0] - (BaseModelConfig.jitter * 2) - (patch_dim*3 + gap*2), image.shape[1] - (BaseModelConfig.jitter * 2) - (patch_dim*3 + gap*2)
        
        start_grid_x, start_grid_y = np.random.randint(0, offset_x) + BaseModelConfig.jitter, np.random.randint(0, offset_y + 1) + BaseModelConfig.jitter
                
        patch_loc_arr = [(1, 1), (1, 2), (1, 3), (2, 1), (2, 3), (3, 1), (3, 2), (3, 3)]
        loc = np.random.randint(len(patch_loc_arr))
        tempx, tempy = patch_loc_arr[loc]
        
        # jitter
        jitter_x = np.random.randint(- BaseModelConfig.jitter, + BaseModelConfig.jitter + 1)
        jitter_y = np.random.randint(- BaseModelConfig.jitter, + BaseModelConfig.jitter + 1)

        patch_x_pt = start_grid_x + patch_dim * (tempx-1) + gap * (tempx-1) + jitter_x
        patch_y_pt = start_grid_y + patch_dim * (tempy-1) + gap * (tempy-1) + jitter_y
        
        random_patch = image[patch_y_pt:patch_y_pt+patch_dim, patch_x_pt:patch_x_pt+patch_dim]

        # jitter
        jitter_x = np.random.randint(- BaseModelConfig.jitter, + BaseModelConfig.jitter + 1)
        jitter_y = np.random.randint(- BaseModelConfig.jitter, + BaseModelConfig.jitter + 1)
        
        patch_x_pt = start_grid_x + patch_dim * (2-1) + gap * (2-1) + jitter_x
        patch_y_pt = start_grid_y + patch_dim * (2-1) + gap * (2-1) + jitter_y
        
        uniform_patch = image[patch_y_pt:patch_y_pt+patch_dim, patch_x_pt:patch_x_pt+patch_dim]

        random_patch_label = loc

        return uniform_patch, random_patch, random_patch_label
    
    def __len__(self):
        return len(self.train_data)
    
    def __getitem__(self, index):
        image = self.train_data[index]
        
        image = np.copy(image)
        
        uniform_patch, random_patch, random_patch_label = self.get_patch_from_grid(image, self.patch_dim, self.gap)

        # save original patches
        uniform_patch_original = copy.deepcopy(uniform_patch)
        random_patch_original = copy.deepcopy(random_patch)   
        
        if uniform_patch.shape[0] != 96:
            uniform_patch = skimage.transform.resize(uniform_patch, (96, 96))
            random_patch = skimage.transform.resize(random_patch, (96, 96))
            uniform_patch_original = skimage.transform.resize(uniform_patch_original, (96, 96))
            random_patch_original = skimage.transform.resize(random_patch_original, (96, 96))

            uniform_patch = img_as_float32(uniform_patch)
            random_patch = img_as_float32(random_patch)
            uniform_patch_original = img_as_float32(uniform_patch_original)
            random_patch_original = img_as_float32(random_patch_original)

            
        # Dropped color channels 2 and 3 and replaced with gaussian noise(std ~1/100 of the std of the remaining channel)
        uniform_patch[:, :, 1] = np.random.normal(0.485, 0.01 * np.std(uniform_patch[:, :, 0]), (uniform_patch.shape[0],uniform_patch.shape[1]))
        uniform_patch[:, :, 2] = np.random.normal(0.485, 0.01 * np.std(uniform_patch[:, :, 0]), (uniform_patch.shape[0],uniform_patch.shape[1]))
        random_patch[:, :, 1] = np.random.normal(0.485, 0.01 * np.std(random_patch[:, :, 0]), (random_patch.shape[0],random_patch.shape[1]))
        random_patch[:, :, 2] = np.random.normal(0.485, 0.01 * np.std(random_patch[:, :, 0]), (random_patch.shape[0],random_patch.shape[1]))

        random_patch_label = np.array(random_patch_label).astype(np.int64)

        if self.transform:
            uniform_patch = self.transform(uniform_patch)
            random_patch = self.transform(random_patch)
            uniform_patch_original = self.transform(uniform_patch_original)
            random_patch_original = self.transform(random_patch_original)
            
        return uniform_patch, random_patch, random_patch_label, uniform_patch_original, random_patch_original
    

##################################################
# Creating Train/Validation dataset and dataloader
##################################################
traindataset = OurDataset(BaseModelConfig.patch_dim, BaseModelConfig.gap, df_trn['image'], False,
                         transforms.Compose([transforms.ToTensor(),
                                             transforms.Normalize(mean=[0.485, 0.456, 0.406], 
                                                                  std=[0.229, 0.224, 0.225])]), 
                          transforms.Compose([transforms.ToTensor()]))
trainloader = torch.utils.data.DataLoader(traindataset, 
                                          batch_size=BaseModelConfig.batch_size,
                                          shuffle=True,
                                          # num_workers=Config.num_workers
                                          )

valdataset = OurDataset(BaseModelConfig.patch_dim, BaseModelConfig.gap, df_val['image'], True,
                         transforms.Compose([transforms.ToTensor(),
                                             transforms.Normalize(mean=[0.485, 0.456, 0.406], 
                                                                  std=[0.229, 0.224, 0.225])]), 
                          transforms.Compose([transforms.ToTensor()]))
valloader = torch.utils.data.DataLoader(valdataset,
                                        batch_size=BaseModelConfig.batch_size,
                                        shuffle=False)
```

```python
##############################
# Visualizing training dataset
##############################

example_batch = next(iter(trainloader))
concatenated = torch.cat((example_batch[3],example_batch[4]),0)
plot_image(torchvision.utils.make_grid(concatenated))
print(f'Labels: {example_batch[2].numpy()}')
```

```python
##############################
# Visualizing validation dataset
##############################

example_batch_val = next(iter(valloader))
concatenated = torch.cat((unorm(example_batch_val[3]),unorm(example_batch_val[4])),0)
plot_image(torchvision.utils.make_grid(concatenated))
print(f'Labels: {example_batch_val[2].numpy()}')
```

```python
class OurNetwork(torch.nn.Module):
    def __init__(self, aux_logits=False):
        super(OurNetwork, self).__init__()
        self.cnn = torch.nn.Sequential(
            torch.nn.Conv2d(3, 96, kernel_size=11, stride=4),
            torch.nn.ReLU(inplace=True),
            torch.nn.MaxPool2d(kernel_size=3, stride=2),
            torch.nn.LocalResponseNorm(96),

            torch.nn.Conv2d(96, 384, kernel_size=5, stride = 2,padding = 2),
            torch.nn.ReLU(inplace=True),
            torch.nn.MaxPool2d(kernel_size=3, stride=2),
            torch.nn.LocalResponseNorm(384),

            torch.nn.Conv2d(384, 384, kernel_size=3, stride=1,padding = 1),
            torch.nn.ReLU(inplace=True),
            torch.nn.BatchNorm2d(384),

            torch.nn.Conv2d(384, 384, kernel_size=3, stride=1,padding = 1),
            torch.nn.ReLU(inplace=True),
            torch.nn.BatchNorm2d(384),

            torch.nn.Conv2d(384, 256, kernel_size=3, stride=1,padding = 1),
            torch.nn.ReLU(inplace=True),
            torch.nn.BatchNorm2d(256),
            torch.nn.MaxPool2d(kernel_size=3, stride=2,padding = 1),
          )
        self.fc6 = torch.nn.Sequential(
            torch.nn.Linear(256,4096),
            torch.nn.ReLU(inplace=True),
            torch.nn.BatchNorm1d(4096),
          )
        self.fc = torch.nn.Sequential(
            torch.nn.Linear(2*4096,4096),
            torch.nn.ReLU(inplace=True),

            torch.nn.Linear(4096, 4096),
            torch.nn.ReLU(inplace=True),

            torch.nn.Linear(4096, 8)
          )

    def forward_once(self, x):
        output= self.cnn(x)
        output = output.view(output.size()[0], -1)
        output = self.fc6(output)
        return output


    def forward(self, uniform_patch, random_patch):
        output_fc6_uniform = self.forward_once(uniform_patch)
        output_fc6_random = self.forward_once(random_patch)
        output = torch.cat((output_fc6_uniform,output_fc6_random), 1)
        output = self.fc(output)
        return output, output_fc6_uniform, output_fc6_random
    
model = OurNetwork().to(device)
print(model)
```

```python

#############################################
# Initialized Optimizer, criterion, scheduler
#############################################

optimizer = optim.Adam(model.parameters(), lr=BaseModelConfig.lr)
criterion = torch.nn.CrossEntropyLoss()
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 
                                           mode='min',
                                           patience=5,
                                           factor=0.3, verbose=True)
```

```python
############################
# Training/Validation Engine
############################
MODEL_SAVE_PATH = f'./models/base_model_{BaseModelConfig.batch_size}_{BaseModelConfig.num_epochs}_{BaseModelConfig.lr}_{BaseModelConfig.patch_dim}_{BaseModelConfig.gap}.pt'


global_trn_loss = []
global_val_loss = []

for epoch in range(BaseModelConfig.num_epochs):
    train_running_loss = []
    val_running_loss = []
    start_time = time.time()
    model.train()
    
    for idx, data in tqdm(enumerate(trainloader), total=int(len(traindataset)/BaseModelConfig.batch_size)):
        uniform_patch, random_patch, random_patch_label = data[0].to(device), data[1].to(device), data[2].to(device)
        optimizer.zero_grad()
        output, output_fc6_uniform, output_fc6_random = model(uniform_patch, random_patch)
        loss = criterion(output, random_patch_label)
        loss.backward()
        optimizer.step()
        
        train_running_loss.append(loss.item())
    correct = 0
    total = 0
    model.eval()
    with torch.no_grad():
        for idx, data in tqdm(enumerate(valloader), total=int(len(valdataset)/BaseModelConfig.batch_size)):
            uniform_patch, random_patch, random_patch_label = data[0].to(device), data[1].to(device), data[2].to(device)
            output, output_fc6_uniform, output_fc6_random = model(uniform_patch, random_patch)
            loss = criterion(output, random_patch_label)
            val_running_loss.append(loss.item())

            _, predicted = torch.max(output.data, 1)
            total += random_patch_label.size(0)
            correct += (predicted == random_patch_label).sum()
    print('Val Progress --- total:{}, correct:{}'.format(total, correct.item()))
    print('Val Accuracy of the network on the 10000 test images: {}%'.format(100 * correct / total))
        
    global_trn_loss.append(sum(train_running_loss) / len(train_running_loss))
    global_val_loss.append(sum(val_running_loss) / len(val_running_loss))
    
    scheduler.step(global_val_loss[-1])
    
    print('Epoch [{}/{}], TRNLoss:{:.4f}, VALLoss:{:.4f}, Time:{:.2f}'.format(
        epoch + 1, BaseModelConfig.num_epochs, global_trn_loss[-1], global_val_loss[-1],
        (time.time() - start_time) / 60))
    
    if epoch % 20 == 0:
        torch.save(
            {
                'epoch': BaseModelConfig.num_epochs,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'loss': loss,
                'global_trnloss': global_trn_loss,
                'global_valloss': global_val_loss
            }, MODEL_SAVE_PATH)
```
